## Train Distribution Check

```python
sns.set_theme(style="whitegrid")

def plot_numeric_distribution(df, col, bins=40):
	fig, axes = plt.subplots(1, 2, figsize=(12, 4))
	sns.histplot(df[col], bins=bins, kde=True, ax=axes[0])
	axes[0].set_title(f"{col} — histogram")
	sns.boxplot(x=df[col], ax=axes[1])
	axes[1].set_title(f"{col} — boxplot")
	plt.tight_layout()
	plt.show()
	print(df[col].describe())

	q1, q3 = df[col].quantile([0.25, 0.75])
	iqr = q3 - q1
	lower, upper = q1 - 1.5 * iqr, q3 + 1.5 * iqr
	outliers = df[col][(df[col] < lower) | (df[col] > upper)]
	print(f"IQR outlier bounds (1.5x rule): [{lower:.2f}, {upper:.2f}]")
	print(f"outliers: {len(outliers)} rows ({len(outliers) / len(df):.2%} of data)")
	if len(outliers):
		print(f"  below lower bound: {(outliers < lower).sum()}, above upper bound: {(outliers > upper).sum()}")

def plot_categorical_distribution(df, col, order=None):
	plt.figure(figsize=(8, 4))
	order = order or df[col].value_counts().index
	ax = sns.countplot(x=df[col], order=order)
	ax.set_title(f"{col} — value counts")
	plt.xticks(rotation=45, ha="right")
	plt.tight_layout()
	plt.show()
	print(df[col].value_counts(normalize=True).round(3))
```

## Multicollinearity (numeric feature-to-feature)

```python
numeric_cols = ["age", "balance", "day", "duration", "campaign", "pdays", "previous"]
categorical_cols = ["job", "marital", "education", "default", "housing", "loan", "contact", "month", "poutcome"]

corr_matrix = train[numeric_cols].corr()
plt.figure(figsize=(8, 6))
sns.heatmap(corr_matrix, annot=True, fmt=".2f", cmap="coolwarm", center=0, vmin=-1, vmax=1)
plt.title("Numeric feature-to-feature correlation")
plt.tight_layout()
plt.show()
```

## Regression EDA — Primary Target: balance

```python
other_numeric = [c for c in numeric_cols if c != "balance"]

balance_corr = train[other_numeric].corrwith(train["balance"]).sort_values(key=abs, ascending=False)
print("balance vs. other numeric features (Pearson r):")
print(balance_corr.round(3))

fig, axes = plt.subplots(2, 3, figsize=(15, 8))
for ax, col in zip(axes.flat, other_numeric):
	sns.scatterplot(x=train[col], y=train["balance"], alpha=0.2, ax=ax)
	ax.set_title(f"balance vs {col} (r={balance_corr[col]:.2f})")
plt.tight_layout()
plt.show()
```

```python
balance_categorical_cols = ["job", "education", "marital", "housing", "loan"]

fig, axes = plt.subplots(len(balance_categorical_cols), 1, figsize=(9, 4 * len(balance_categorical_cols)))
for ax, col in zip(axes, balance_categorical_cols):
	order = train.groupby(col)["balance"].median().sort_values(ascending=False).index
	sns.boxplot(x=train[col], y=train["balance"], order=order, ax=ax, showfliers=False)
	ax.set_title(f"balance by {col} (outliers hidden for readability)")
	ax.tick_params(axis="x", rotation=45)
plt.tight_layout()
plt.show()
```

## Model-Based Feature Importance (RF)

```python
def build_feature_matrix(df, exclude=None):
	exclude = set(exclude or [])
	num_cols = [c for c in numeric_cols if c not in exclude]
	cat_cols = [c for c in categorical_cols if c not in exclude]
	frames = [df[num_cols]]
	col_to_parent = {c: c for c in num_cols}
	for col in cat_cols:
		dummies = pd.get_dummies(df[col], prefix=col)
		frames.append(dummies)
		col_to_parent.update({d: col for d in dummies.columns})
	return pd.concat(frames, axis=1), col_to_parent

X_reg, col_to_parent_reg = build_feature_matrix(train, exclude=["balance"])
rf_reg = RandomForestRegressor(n_estimators=300, random_state=42, n_jobs=-1)
rf_reg.fit(X_reg, train["balance"])

importances_reg = pd.Series(rf_reg.feature_importances_, index=X_reg.columns)
importances_reg_agg = importances_reg.groupby(col_to_parent_reg).sum().sort_values(ascending=False)
importances_reg_agg
```

```python
plt.figure(figsize=(8, 6))
sns.barplot(x=importances_reg_agg.values, y=importances_reg_agg.index, hue=importances_reg_agg.index,
            palette="Greens_r", legend=False)
plt.title("RF feature importance — balance (regression)")
plt.xlabel("aggregated importance")
plt.tight_layout()
plt.show()
```

## Relevance Cross-Check — Mutual Information

```python
from sklearn.feature_selection import mutual_info_regression

mi_numeric = pd.Series(
	mutual_info_regression(train[other_numeric], train["balance"], discrete_features=False, random_state=42),
	index=other_numeric,
)

cat_codes = train[categorical_cols].apply(lambda s: pd.factorize(s)[0])
mi_categorical = pd.Series(
	mutual_info_regression(cat_codes, train["balance"], discrete_features=True, random_state=42),
	index=categorical_cols,
)

mi_relevance = pd.concat([mi_numeric, mi_categorical]).sort_values(ascending=False)
mi_relevance_df = mi_relevance.rename("mutual_info").to_frame()
mi_relevance_df["type"] = ["numeric" if c in other_numeric else "categorical" for c in mi_relevance_df.index]
mi_relevance_df
```

```python
plt.figure(figsize=(8, 6))
sns.barplot(x=mi_relevance.values, y=mi_relevance.index, hue=mi_relevance.index,
            palette="Purples_r", legend=False)
plt.title("Mutual information — balance (regression)")
plt.xlabel("mutual information")
plt.tight_layout()
plt.show()
```
