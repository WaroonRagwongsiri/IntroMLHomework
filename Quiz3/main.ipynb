```python
import pandas as pd
import numpy as np
import torch
import sklearn
```

```python
# Class labels
labels = [
    'T-Shirt', 'Trouser', 'Pullover', 'Dress', 'Coat',
    'Sandal', 'Shirt', 'Sneaker', 'Bag', 'Ankle boot'
]

# Confusion matrix data array (Rows = REAL / True, Columns = PRED / Predicted)
cm_data = np.array([
    [783,   0,  15,  12,   2,   1,  73,   0,   6,   0],  # T-Shirt
    [  0, 876,   2,   8,   1,   0,   3,   0,   0,   0],  # Trouser
    [ 12,   1, 811,   8,  35,   0,  38,   0,   0,   0],  # Pullover
    [ 20,   4,  10, 836,  21,   0,  24,   0,   1,   0],  # Dress
    [  0,   0,  65,  20, 776,   0,  52,   0,   0,   0],  # Coat
    [  0,   0,   0,   0,   0, 884,   0,  10,   0,   6],  # Sandal
    [ 88,   1,  44,  19,  50,   0, 683,   0,   4,   0],  # Shirt
    [  0,   0,   0,   0,   0,   5,   0, 888,   0,  17],  # Sneaker
    [  4,   1,   1,   1,   4,   2,   3,   1, 868,   1],  # Bag
    [  0,   0,   0,   0,   0,   4,   0,  18,   1, 876]   # Ankle boot
])

# Pandas DataFrame format
cm_df = pd.DataFrame(cm_data, index=labels, columns=labels)
cm_df.index.name = 'REAL'
cm_df.columns.name = 'PRED'

# Display DataFrame
cm_df
```

```python
total = cm_data.sum()
tp = np.diag(cm_data)
fn = cm_data.sum(axis=1) - tp
fp = cm_data.sum(axis=0) - tp
tn = total - (tp + fp + fn)

precision = tp / (tp + fp)
recall = tp / (tp + fn)
specificity = tn / (tn + fp)
f1 = 2 * (precision * recall) / (precision + recall)
accuracy_per_class = (tp + tn) / total

metrics_df = pd.DataFrame({
    'TP': tp,
    'FP': fp,
    'FN': fn,
    'TN': tn,
    'Precision': precision,
    'Recall (Sensitivity)': recall,
    'Specificity': specificity,
    'F1-Score': f1
}, index=labels)

overall_acc = tp.sum() / total

print("Overall Accuracy:", overall_acc)
print("\nMetrics table:")
print(metrics_df.round(4))
```
